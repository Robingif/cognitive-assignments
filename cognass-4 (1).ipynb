```python
import pandas as pd

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    },
    {
        "question": "how do i update my registered mobile number",
        "answer": "You can update your registered mobile number from account settings.",
        "keywords": "mobile update phone account",
        "category": "account"
    },
    {
        "question": "how do i change my account email",
        "answer": "You can change your account email from profile settings.",
        "keywords": "email update account profile",
        "category": "account"
    }
]

df = pd.DataFrame(fixed_entries)

print(df)
```

```python
def score_hypothesis(query):
    query = query.lower()
    
    results = []

    for index, row in df.iterrows():
        score = 0
        
        # Combine question and keywords for matching
        text = (row["question"] + " " + row["keywords"]).lower()
        
        # Split query words
        query_words = query.split()
        
        # Count matching words
        for word in query_words:
            if word in text:
                score += 1
        
        # Convert score into confidence percentage
        confidence = (score / len(query_words)) * 100
        
        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": round(confidence, 2)
            })

    # Rank results by confidence
    results = sorted(results, key=lambda x: x["confidence"], reverse=True)

    return results


# Testing the scoring function
query = "how do i update my mobile number"

matches = score_hypothesis(query)

for result in matches:
    print(result)
    
```

```python
def same_category(category_name, df):
    result=[]
    for index,row in df.iterrows():
        
        if row["category"]==category_name:
            result.append(row["question"])

    return result
category = "account"

questions = same_category(category, df)

print(questions)
            
    
```

```python
def score_hypothesis(query):
    query_words = query.lower().split()

    hypotheses = []

    for index, row in df.iterrows():

        score = 0

        keywords = row["keywords"].lower().split()

        for word in query_words:
            if word in keywords:
                score += 1

        hypotheses.append({
            "question": row["question"],
            "answer": row["answer"],
            "category": row["category"],
            "score": score
        })


    # Find highest score
    highest_score = max(h["score"] for h in hypotheses)

    # Store all entries having highest score
    best_matches = []

    for h in hypotheses:
        if h["score"] == highest_score:
            best_matches.append(h)

    return best_matches
```

```python
result = score_hypothesis("fee")

for r in result:
    print(r)
```
