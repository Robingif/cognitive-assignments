```python
 x = np.array([1,2,3,4,5,1,2,1,1,1]) 
 y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ]) 
values, counts = np.unique(x, return_counts=True)

max_value = values[np.argmax(counts)]

indices = np.where(x == max_value)[0]
print("most frequent in x:",max_value)
print("at",indices)
```

```python
values, counts = np.unique(y, return_counts=True)

max_value = values[np.argmax(counts)]

indices = np.where(y == max_value)[0]
print("most frequent in y:",max_value)
print("at",indices)
```

```python
robin = np.linspace(10,100,25)
print(robin)
```

```python
print(robin.size)
print(robin.shape)
print(robin.dtype)
print(robin.ndim)
print("Total bytes:", robin.nbytes)
robin_2D = robin.reshape(5,5)

print("\n2-D Array:")
print(robin_2D)

print("\nTranspose using reshape:")
print(robin_2D.reshape(5,5).T)
```

```python
ucs420_robin=np.array([[10,20,30,40],[50,60,70,80],[90,15,20,35]])
print(np.mean(ucs420_robin))
print(np.median(ucs420_robin))
print(int(np.max(ucs420_robin)))
print(int(np.min(ucs420_robin)))
print(np.unique(ucs420_robin))

```

```python
reshaped_ucs420_robin=ucs420_robin.reshape(4,3)
print(reshaped_ucs420_robin)
```
